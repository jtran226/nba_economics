### Getting all players who got NBA first, second, third team from BR and bringing it into one nice dataframe...

```python
import pandas as pd
import os
import certifi 
import requests

from io import StringIO

all_nba_url = "https://www.basketball-reference.com/awards/all_league.html"

response = requests.get(
    all_nba_url,
    verify=certifi.where(),
    timeout=20
)

response.raise_for_status()

all_nba_tables = pd.read_html(StringIO(response.content.decode("utf-8")))

len(all_nba_tables)
```

```python
all_nba = all_nba_tables[0]

all_nba.shape
```

```python
all_nba.head(25)
```

```python
all_nba.info()
```

```python
all_nba_clean1 = all_nba[all_nba["Season"].notna()].copy()

all_nba_clean1.head(25)
```

```python
new_column_names = ["season", "league", "team", "vote", "center", "power_forward", "small_forward", "shooting_guard", "point_guard"]

all_nba_clean1.columns = new_column_names

all_nba_clean1.head()
```

```python
all_nba_clean2 = all_nba_clean1[["season", "team", "center", "power_forward", "small_forward", "shooting_guard", "point_guard"]]

all_nba_clean2.head()
```

```python
all_nba_long = pd.melt(all_nba_clean2,
id_vars = ["season", "team"],
value_vars = ["center", "power_forward", "small_forward", "shooting_guard", "point_guard"],
var_name = "position",
value_name = "player_name")

all_nba_long.head(30)
```

```python
all_nba_long = all_nba_long[all_nba_long["season"].str[:4].astype(int).between(2000,2025)].copy()

all_nba_long.shape
```

```python
all_nba_long.tail()
```

```python
all_nba_long["source_slot"] = all_nba_long["position"].map({
    "center": 1,
    "power_forward": 2,
    "small_forward": 3,
    "shooting_guard": 4,
    "point_guard": 5
})

all_nba_long["player_position"] = (
    all_nba_long["player_name"].str.extract(r"\s([CFG])$")
)

all_nba_long["player_name"] = (
    all_nba_long["player_name"]
    .str.replace(r"\s+[CFG]$", "", regex=True)
    .str.strip()
)

all_nba_long.head()
```

```python
all_nba_long["season"] = (
    all_nba_long["season"].str[:5]
    + "20"
    + all_nba_long["season"].str[-2:]
)

all_nba_long["all_nba_team"] = (
    all_nba_long["team"].map({
        "1st": 1,
        "2nd": 2,
        "3rd": 3
    })
)

all_nba_long["award_type"] = "All-NBA"

all_nba_long.head(25)

```

```python
all_nba_final = all_nba_long[
    ["season", "all_nba_team", "player_name", "player_position"]
].copy()

all_nba_final.shape
```

```python
all_nba_final.head()
```

```python
all_nba_final.info()
```

```python
staging_dir = "data/staging/awards/S14A_all_nba"

os.makedirs(staging_dir, exist_ok=True)
```

```python
all_nba_final.to_csv(
    f"{staging_dir}/S14A_all_nba.csv",
    index=False,
    encoding="utf-8"
)
```
