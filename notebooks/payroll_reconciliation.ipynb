```python
import pandas as pd
from pathlib import Path

# Define the project directories.
project_root = Path.cwd().parent
staging_dir = project_root / "data" / "staging"
canonical_dir = project_root / "data" / "canonical"

# Load S08 (RealGM), S18 (Spotrac), and C04 (canonical team seasons).
s08 = pd.read_csv(staging_dir / "team_payroll" / "s08_team_payroll.csv")
s18 = pd.read_csv(staging_dir / "luxury_tax" / "S18_Luxury_Tax.csv")
c04 = pd.read_csv(canonical_dir / "team_seasons.csv")

# Inspect dataset size, column names, and sample records.
for name, df in {
    "S08 — RealGM team payroll": s08,
    "S18 — Spotrac luxury tax": s18,
    "C04 — Canonical team seasons": c04
}.items():
    print(f"\n{'=' * 60}")
    print(name)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("\nFirst 3 rows:")
    print(df.head(3).to_string(index=False))
```

```python
# Locate the canonical franchise lookup table.
franchises = pd.read_csv(canonical_dir / "franchises.csv")

# Inspect its structure and a few representative records.
print("Columns:", franchises.columns.tolist())
print("\nSample records:")
print(franchises.head(10).to_string(index=False))
```

```python
# Create a lookup from three-letter abbreviation to numeric NBA franchise ID.
franchise_lookup = franchises[
    ["franchise_id", "current_abbreviation"]
].rename(columns={
    "franchise_id": "canonical_franchise_id"
})

# Join S18 to the lookup using the abbreviation.
s18_mapped = s18.merge(
    franchise_lookup,
    left_on="franchise_id",
    right_on="current_abbreviation",
    how="left",
    validate="many_to_one"
)

# Check how many Spotrac records received a numeric franchise ID.
print("Total S18 rows:", len(s18_mapped))
print("Mapped rows:", s18_mapped["canonical_franchise_id"].notna().sum())
print("Unmapped rows:", s18_mapped["canonical_franchise_id"].isna().sum())

# Identify abbreviations that did not match the current lookup.
print("\nUnmapped abbreviations:")
print(
    s18_mapped.loc[
        s18_mapped["canonical_franchise_id"].isna(),
        "franchise_id"
    ].value_counts()
)
```

```python
# Define the three known differences between S18 and canonical abbreviations.
abbreviation_map = {
    "BRK": "BKN",
    "PHO": "PHX",
    "CHO": "CHA"
}

# Create a normalized abbreviation column.
# replace() changes only the three specified values.
s18["normalized_abbreviation"] = s18["franchise_id"].replace(abbreviation_map)

# Join against our existing franchise lookup.
s18_mapped = s18.merge(
    franchise_lookup,
    left_on="normalized_abbreviation",
    right_on="current_abbreviation",
    how="left",
    validate="many_to_one"
)

# Verify that all 718 records received a canonical franchise ID.
print("Total S18 rows:", len(s18_mapped))
print("Mapped rows:", s18_mapped["canonical_franchise_id"].notna().sum())
print("Unmapped rows:", s18_mapped["canonical_franchise_id"].isna().sum())
```

```python
# Define source-to-canonical column mappings.
financial_map = {
    "tax_threshold": "s18_tax_threshold",
    "taxable_payroll": "taxable_payroll",
    "tax_space": "tax_space",
    "tax_paid": "tax_paid",
    "estimated_tax_bill": "estimated_tax_bill",
    "total_estimated_cost": "total_estimated_cost",
    "active_salary": "active_salary",
    "dead_salary": "dead_salary"
}

# Join mapped S18 rows to their canonical franchise-season records.
comparison = s18_mapped.merge(
    c04[["franchise_id", "season"] + list(financial_map.values())],
    left_on=["canonical_franchise_id", "season"],
    right_on=["franchise_id", "season"],
    how="left",
    suffixes=("_s18", "_c04"),
    indicator=True,
    validate="one_to_one"
)

print("\nJoin results:")
print(comparison["_merge"].value_counts())

# Compare every financial field, counting missing values on both sides as equal.
for source_col, canonical_col in financial_map.items():
    left = comparison[f"{source_col}_s18"] if source_col != "tax_threshold" else comparison["tax_threshold"]
    right = comparison[f"{canonical_col}_c04"] if canonical_col != "s18_tax_threshold" else comparison["s18_tax_threshold"]

    mismatches = ~(left.eq(right) | (left.isna() & right.isna()))
    print(f"{source_col}: {mismatches.sum()} mismatches")
```

```python
# Compare the distinct team names in RealGM and canonical C04.
s08_names = set(s08["Team_Name"].dropna().unique())
c04_names = set(c04["team_name"].dropna().unique())

print("Unique S08 names:", len(s08_names))
print("Unique C04 names:", len(c04_names))

print("\nNames only in S08:")
print(sorted(s08_names - c04_names))

print("\nNames only in C04:")
print(sorted(c04_names - s08_names))
```

```python
# Load the historical franchise-season lookup created during Phase A.
aliases = pd.read_csv(
    canonical_dir / "franchise_season_aliases.csv"
)

# Inspect the available identifiers and naming conventions.
print("Shape:", aliases.shape)
print("Columns:", aliases.columns.tolist())

print("\nSample rows:")
print(aliases.head(10).to_string(index=False))
```

```python
# STEP A: Map RealGM's shortened team names to canonical franchise IDs.
s08_mapped = s08.merge(
    aliases[
        ["franchise_id", "season", "short_team_name"]
    ],
    left_on=["Team_Name", "Season"],
    right_on=["short_team_name", "season"],
    how="left",
    validate="one_to_one",
    indicator=True
)

print("S08 → alias mapping:")
print(s08_mapped["_merge"].value_counts())

# Show any RealGM records that did not find an alias.
print("\nUnmatched RealGM records:")
print(
    s08_mapped.loc[
        s08_mapped["_merge"] == "left_only",
        ["Team_Name", "Season"]
    ].to_string(index=False)
)

# STEP B: Compare the mapped RealGM records with canonical C04.
payroll_map = {
    "Team_Payroll": "team_payroll",
    "Salary_Cap_Hit": "salary_cap_hit",
    "Salary_Cap_Space": "salary_cap_space"
}

s08_comparison = s08_mapped.merge(
    c04[
        ["franchise_id", "season"] + list(payroll_map.values())
    ],
    on=["franchise_id", "season"],
    how="left",
    validate="one_to_one",
    indicator="canonical_match"
)

print("\nS08 → C04 join:")
print(s08_comparison["canonical_match"].value_counts())

# Compare values, treating two missing values as equal.
for source_col, canonical_col in payroll_map.items():
    left = s08_comparison[source_col]
    right = s08_comparison[canonical_col]

    mismatches = ~(left.eq(right) | (left.isna() & right.isna()))

    print(f"{source_col}: {mismatches.sum()} mismatches")
```

```python
# Select the 2024–25 season from canonical C04.
sample = c04.loc[
    c04["season"] == "2024-2025",
    [
        "franchise_id",
        "team_name",
        "season",
        "team_payroll",
        "taxable_payroll",
        "s18_tax_threshold",
        "tax_paid"
    ]
].copy()

# Calculate the dollar difference between the two payroll measures.
# Positive values mean RealGM payroll exceeds Spotrac taxable payroll.
sample["payroll_difference"] = (
    sample["team_payroll"] - sample["taxable_payroll"]
)

# Express the difference relative to Spotrac taxable payroll.
sample["difference_pct"] = (
    sample["payroll_difference"] / sample["taxable_payroll"] * 100
)

# Sort by largest absolute dollar differences.
sample = sample.sort_values(
    "payroll_difference",
    key=lambda x: x.abs(),
    ascending=False
)

# Format financial columns for readability without changing the underlying data.
print(
    sample[
        [
            "team_name",
            "team_payroll",
            "taxable_payroll",
            "payroll_difference",
            "difference_pct"
        ]
    ].to_string(
        index=False,
        formatters={
            "team_payroll": "{:,.0f}".format,
            "taxable_payroll": "{:,.0f}".format,
            "payroll_difference": "{:+,.0f}".format,
            "difference_pct": "{:+.1f}%".format
        }
    )
)
```

```python
# Check that every franchise has the necessary financial data.
cols_to_check = [
    "team_payroll",
    "taxable_payroll",
    "s18_tax_threshold"
]

print("Missing values:")
print(sample[cols_to_check].isna().sum())

# Confirm whether Spotrac's tax threshold is consistent across teams.
print("\nUnique luxury-tax thresholds:")
print(sample["s18_tax_threshold"].dropna().unique())

# Summarize the size of payroll discrepancies.
print("\nPayroll difference summary (USD millions):")
print(
    (sample["payroll_difference"].abs() / 1_000_000)
    .describe()
    .round(2)
)
```

```python
# Identify whether each payroll measure exceeds the luxury-tax threshold.
# These are analytical proxies, NOT verified official tax classifications.
sample["above_tax_realgm_proxy"] = (
    sample["team_payroll"] > sample["s18_tax_threshold"]
)

sample["above_tax_spotrac_proxy"] = (
    sample["taxable_payroll"] > sample["s18_tax_threshold"]
)

# Flag franchises where the two classifications disagree.
sample["tax_classification_disagreement"] = (
    sample["above_tax_realgm_proxy"]
    != sample["above_tax_spotrac_proxy"]
)

# Summarize the comparison.
print("2024–25 franchise-seasons:", len(sample))
print("RealGM payroll above tax threshold:",
      sample["above_tax_realgm_proxy"].sum())
print("Spotrac taxable payroll above tax threshold:",
      sample["above_tax_spotrac_proxy"].sum())

print("\nFranchises with different classifications:")
print(
    sample.loc[
        sample["tax_classification_disagreement"],
        [
            "team_name",
            "team_payroll",
            "taxable_payroll",
            "s18_tax_threshold"
        ]
    ].to_string(index=False)
)
```

```python
# Official published 2024–25 second-apron threshold.
second_apron = 188_931_000

# Calculate each franchise's distance from the second apron
# using both payroll measures as preliminary proxies.
sample["realgm_second_apron_space"] = (
    second_apron - sample["team_payroll"]
)

sample["spotrac_second_apron_space"] = (
    second_apron - sample["taxable_payroll"]
)

# Compare the resulting above-apron classifications.
sample["realgm_above_second_apron"] = (
    sample["team_payroll"] > second_apron
)

sample["spotrac_above_second_apron"] = (
    sample["taxable_payroll"] > second_apron
)

print("RealGM above second apron:",
      sample["realgm_above_second_apron"].sum())

print("Spotrac above second apron:",
      sample["spotrac_above_second_apron"].sum())

# Display any franchises where the classifications differ.
print("\nClassification disagreements:")
print(
    sample.loc[
        sample["realgm_above_second_apron"]
        != sample["spotrac_above_second_apron"],
        [
            "team_name",
            "team_payroll",
            "taxable_payroll",
            "realgm_second_apron_space",
            "spotrac_second_apron_space"
        ]
    ].to_string(index=False)
)
```

```python
# Work with our existing 2024–25 sample.
tax_test = sample[
    [
        "team_name",
        "team_payroll",
        "taxable_payroll",
        "s18_tax_threshold",
        "tax_paid"
    ]
].copy()

# Calculate how far Spotrac taxable payroll sits above/below the tax line.
tax_test["tax_margin"] = (
    tax_test["taxable_payroll"] - tax_test["s18_tax_threshold"]
)

# Create a preliminary tax classification using taxable payroll.
tax_test["above_tax_proxy"] = tax_test["tax_margin"] > 0

# Check whether Spotrac records a positive tax payment.
# Missing tax payments are kept separate from confirmed zero payments.
tax_test["tax_payment_recorded"] = tax_test["tax_paid"] > 0

# Only compare classifications where tax_paid is actually populated.
tax_test["classification_agrees"] = (
    tax_test["above_tax_proxy"] == tax_test["tax_payment_recorded"]
)

print("Teams above tax using taxable payroll:",
      tax_test["above_tax_proxy"].sum())

print("Teams with positive recorded tax payments:",
      tax_test["tax_payment_recorded"].sum())

print("Missing tax_paid values:",
      tax_test["tax_paid"].isna().sum())

# Inspect disagreements only where the tax payment is known.
disagreements = tax_test.loc[
    tax_test["tax_paid"].notna()
    & ~tax_test["classification_agrees"]
]

print("\nDisagreements:")
print(
    disagreements[
        ["team_name", "taxable_payroll", "tax_margin", "tax_paid"]
    ].to_string(index=False)
)
```

```python
import numpy as np

# Preserve the original source fields in a separate working table.
tax_features = c04[
    [
        "franchise_id",
        "season",
        "tax_paid",
        "estimated_tax_bill",
        "taxable_payroll",
        "s18_tax_threshold"
    ]
].copy()

# Establish whether a team has sufficient data to evaluate tax exposure.
has_tax_data = (
    tax_features["taxable_payroll"].notna()
    & tax_features["s18_tax_threshold"].notna()
)

# Identify below-threshold franchises using Spotrac's taxable payroll.
below_tax = (
    has_tax_data
    & (tax_features["taxable_payroll"] <= tax_features["s18_tax_threshold"])
)

# Begin with historical payments, then use estimates where available.
tax_features["luxury_tax_burden"] = (
    tax_features["tax_paid"]
    .combine_first(tax_features["estimated_tax_bill"])
)

# Assign zero only where the payroll and threshold establish
# a below-threshold classification and no amount was reported.
infer_zero = (
    tax_features["luxury_tax_burden"].isna()
    & below_tax
)

tax_features.loc[infer_zero, "luxury_tax_burden"] = 0

# Track how each value was obtained.
tax_features["tax_burden_type"] = np.select(
    [
        tax_features["tax_paid"].notna(),
        tax_features["estimated_tax_bill"].notna(),
        infer_zero
    ],
    [
        "reported_historical",
        "estimated",
        "inferred_zero"
    ],
    default="unknown"
)

print(tax_features["tax_burden_type"].value_counts())
```

```python
# Published 2024–25 apron payrolls from Eric Pincus's
# Sports Business Classroom tracker (updated April 22, 2025).
apron_reference = {
    "Atlanta Hawks": 170381400,
    "Boston Celtics": 193348445,
    "Brooklyn Nets": 176066723,
    "Charlotte Hornets": 167495247,
    "Chicago Bulls": 168731854,
    "Cleveland Cavaliers": 172922353,
    "Dallas Mavericks": 178128841,
    "Denver Nuggets": 183774315,
    "Detroit Pistons": 142808943,
    "Golden State Warriors": 177887433,
    "Houston Rockets": 170921311,
    "Indiana Pacers": 171600105,
    "LA Clippers": 170720454,
    "Los Angeles Lakers": 188260871,
    "Memphis Grizzlies": 166894399,
    "Miami Heat": 176170988,
    "Milwaukee Bucks": 185621334,
    "Minnesota Timberwolves": 205543620,
    "New Orleans Pelicans": 172790840,
    "New York Knicks": 188877650,
    "Oklahoma City Thunder": 167068467,
    "Orlando Magic": 152959238,
    "Philadelphia 76ers": 171079692,
    "Phoenix Suns": 214871340,
    "Portland Trail Blazers": 168712097,
    "Sacramento Kings": 172343977,
    "San Antonio Spurs": 167638808,
    "Toronto Raptors": 176868326,
    "Utah Jazz": 154923317,
    "Washington Wizards": 177224214
}

# Convert the team-to-payroll dictionary into a two-column DataFrame.
apron_benchmark = pd.DataFrame(
    apron_reference.items(),
    columns=["team_name", "apron_payroll_reference"]
)

# Select the matching season and relevant canonical fields.
apron_comparison = c04.loc[
    c04["season"] == "2024-2025",
    ["team_name", "taxable_payroll"]
].merge(
    apron_benchmark,
    on="team_name",
    how="outer",
    validate="one_to_one",
    indicator=True
)

# Confirm all teams joined correctly before evaluating differences.
print("Join results:")
print(apron_comparison["_merge"].value_counts())

# Calculate the dollar difference between the two accounting measures.
apron_comparison["payroll_difference"] = (
    apron_comparison["apron_payroll_reference"]
    - apron_comparison["taxable_payroll"]
)

# Official 2024–25 apron thresholds.
first_apron = 178132000
second_apron = 188931000

# Compare above-threshold classifications from both sources.
for level, threshold in [
    ("first", first_apron),
    ("second", second_apron)
]:
    apron_comparison[f"spotrac_above_{level}"] = (
        apron_comparison["taxable_payroll"] > threshold
    )
    apron_comparison[f"reference_above_{level}"] = (
        apron_comparison["apron_payroll_reference"] > threshold
    )

# Report any disagreements in first- or second-apron classification.
disagreements = apron_comparison.loc[
    (apron_comparison["spotrac_above_first"]
     != apron_comparison["reference_above_first"])
    |
    (apron_comparison["spotrac_above_second"]
     != apron_comparison["reference_above_second"])
]

print("\nClassification disagreements:")
print(disagreements.to_string(index=False))

print("\nLargest absolute payroll differences:")
print(
    apron_comparison.reindex(
        apron_comparison["payroll_difference"].abs()
        .sort_values(ascending=False).index
    )[
        ["team_name", "taxable_payroll",
         "apron_payroll_reference", "payroll_difference"]
    ].head(10).to_string(index=False)
)
```
