```python
from nba_api.stats.endpoints import leaguedashplayerbiostats

def get_roster(year):
    season= f"{year - 1}-{str(year)[-2:]}"

    roster = leaguedashplayerbiostats.LeagueDashPlayerBioStats(
    season = season)

    roster_df = roster.get_data_frames()[0]

    roster_df["season"] = f"{year - 1}-{year}"
    roster_df["year"] = year

    return roster_df

seasons_of_rosters = []

for year in range(2001, 2027):

    roster_df = get_roster(year)
    seasons_of_rosters.append(roster_df)
```

```python
len(seasons_of_rosters)
```

```python
import pandas as pd
import numpy as np

whole_roster = pd.concat(seasons_of_rosters, ignore_index=True)

print(whole_roster.shape)
print(whole_roster.columns)
```

```python
whole_roster.isna().sum()
```

```python
import os

os.makedirs("../data/staging/rosters", exist_ok=True)

whole_roster.to_csv(
    "../data/staging/rosters/s04_rosters.csv",
    index=False
)
```
