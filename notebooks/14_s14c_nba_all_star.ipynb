```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from io import StringIO
import time

# Initialize an empty dictionary
raw_all_stars = {}

for year in range(2001, 2027):

    print(f"Extracting {year}...")

    url = f"https://en.wikipedia.org/wiki/{year}_NBA_All-Star_Game"

    try:
        response = requests.get(
            url,
            headers={"User-Agent": "NBAEconomicsResearch/0.1 (educational project)"},
            timeout=20
        )
        response.raise_for_status()

        soup = BeautifulSoup(response.content, "html.parser")

        # Repair malformed HTML attributes
        for cell in soup.find_all(["td", "th"]):
            for attr in ["rowspan", "colspan"]:
                if cell.has_attr(attr):
                    cell[attr] = cell[attr].strip().rstrip(";")

        # Read all HTML tables
        tables = pd.read_html(StringIO(str(soup)))

        # Create an empty list for this year
        raw_all_stars[year] = []

        # Find the All-Star roster tables
        for i, df in enumerate(tables):

            columns = [
                str(col[0] if isinstance(col, tuple) else col)
                .strip().lower()
                for col in df.columns
            ]

            has_player = "player" in columns
            has_team = "team" in columns

            has_selections = any(
                "selection" in col or "appearance" in col
                for col in columns
            )

            if has_player and has_team and has_selections:
                raw_all_stars[year].append(df.copy())

        print(f"  Stored {len(raw_all_stars[year])} roster tables")

    except Exception as e:
        print(f"  ERROR: {e}")

    time.sleep(2)
```

```python

# S14C — Identify conference from historical NBA team names

EAST_TEAMS = {
    "Atlanta Hawks", "Boston Celtics", "Brooklyn Nets",
    "Charlotte Bobcats", "Charlotte Hornets",
    "Chicago Bulls", "Cleveland Cavaliers",
    "Detroit Pistons", "Indiana Pacers", "Miami Heat",
    "Milwaukee Bucks", "New Jersey Nets", "New York Knicks",
    "Orlando Magic", "Philadelphia 76ers",
    "Toronto Raptors", "Washington Wizards"
}

WEST_TEAMS = {
    "Dallas Mavericks", "Denver Nuggets",
    "Golden State Warriors", "Houston Rockets",
    "Los Angeles Clippers", "Los Angeles Lakers",
    "Memphis Grizzlies", "Minnesota Timberwolves",
    "New Orleans Pelicans", "Oklahoma City Thunder",
    "Phoenix Suns", "Portland Trail Blazers",
    "Sacramento Kings", "San Antonio Spurs",
    "Seattle SuperSonics", "Utah Jazz",
    "Vancouver Grizzlies",
    "New Orleans/Oklahoma City Hornets"
}



def identify_conference(raw_df, year):

    # Read team names from the third column
    teams = (
        raw_df.iloc[:, 2]
        .dropna()
        .astype(str)
        .str.strip()
    )

    # Remove section headers and coaching information
    teams = teams[
        ~teams.str.contains(
            r"Reserves|Head coach",
            case=False,
            regex=True,
            na=False
        )
    ]

    # Remove Wikipedia footnote annotations
    teams = (
        teams
        .str.replace(r"NOTE\d+", "", regex=True)
        .str.replace(r"\[\d+\]", "", regex=True)
        .str.strip()
    )

    east = 0
    west = 0
    unknown = set()

    for team in teams:

        # Historical New Orleans conference switch
        if team == "New Orleans Hornets":
            if year <= 2004:
                east += 1
            else:
                west += 1

        elif team in EAST_TEAMS:
            east += 1

        elif team in WEST_TEAMS:
            west += 1

        else:
            unknown.add(team)

    if unknown:
        raise ValueError(
            f"{year}: Unrecognized teams: {sorted(unknown)}"
        )

    # Determine conference by majority of roster players
    total = east + west

    if total == 0:
        raise ValueError(f"{year}: No recognized NBA teams")

    if east / total >= 0.75:
        return "east"

    elif west / total >= 0.75:
        return "west"

    else:
        raise ValueError(
            f"{year}: Ambiguous conference "
            f"(East={east}, West={west})"
    )
```

```python

def clean_all_star(raw_df, year, conference):
    df = raw_df.copy()

    # 1. Standardize columns
    if df.shape[1] == 5:
        df.columns = [
            "position", "player_name", "team",
            "all_star_appearances", "votes"
        ]
    elif df.shape[1] == 4:
        df.columns = [
            "position", "player_name", "team",
            "all_star_appearances"
        ]
    else:
        raise ValueError(f"Unexpected column count: {df.shape[1]}")

    # 2. Identify starters and reserves
    reserve_index = df[
        df["position"].astype(str).str.contains("Reserves", na=False)
    ].index[0]

    df.loc[df.index < reserve_index, "selection_type"] = "starter"
    df.loc[df.index > reserve_index, "selection_type"] = "reserve"

    # 3. Normalize positions and retain valid player rows
    df["position"] = (
        df["position"]
        .astype(str)
        .str.upper()
        .str.strip()
        .str.replace(r"\[\d+\]", "", regex=True)
        .str.replace(r"\d+", "", regex=True)
        .str.replace(r"[^GFC]", "", regex=True)
    )

    valid_positions = {
        "G", "F", "C",
        "GF", "FG",
        "FC", "CF",
        "GC", "CG"
    }

    df = df[
        df["position"].isin(valid_positions)
    ].copy()

    # Restore readable position labels
    df["position"] = df["position"].apply(
        lambda x: "/".join(x) if len(x) == 2 else x
    )

    # 4. Extract annotations and clean names
    df["player_status"] = df["player_name"].str.extract(
        r"(INJ|REP|SPL)", expand=False
    )

    df["player_name"] = (
        df["player_name"]
        .str.replace(r"(INJ|REP|SPL)", "", regex=True)
        .str.replace(r"[\d\s]+$", "", regex=True)
        .str.strip()
    )

    # 5. Convert data types
    df["all_star_appearances"] = (
        df["all_star_appearances"]
        .astype(str)
        .str.extract(r"(\d+)", expand=False)
        .astype("Int64")
    )

    if "votes" in df.columns:
        df["votes"] = pd.to_numeric(
            df["votes"], errors="coerce"
        ).astype("Int64")

    # 6. Add metadata
    df["year"] = year
    df["season"] = f"{year - 1}-{year}"
    df["conference"] = conference

    # 7. Reset index
    df = df.reset_index(drop=True)

    return df


# Quick test using 2008 Eastern Conference
test_2008 = clean_all_star(
    raw_all_stars[2008][1],
    2008,
    "east"
)

display(test_2008.head())
```

```python

# S14C — Clean historical All-Star rosters, 2001–2026

cleaned_all_stars = []
failed_all_stars = []

for year in range(2001, 2027):

    tables = raw_all_stars.get(year, [])

    # Every year should have two conference rosters
    if len(tables) != 2:
        failed_all_stars.append({
            "year": year,
            "table": None,
            "error": f"Expected 2 tables, found {len(tables)}"
        })
        continue

    year_results = []
    year_failed = False

    for i, raw_df in enumerate(tables):

        try:
            conference = identify_conference(raw_df, year)

            cleaned_df = clean_all_star(
                raw_df,
                year,
                conference
            )

            year_results.append(cleaned_df)

        except Exception as e:
            year_failed = True

            failed_all_stars.append({
                "year": year,
                "table": i,
                "error": str(e)
            })

    # Accept the year only if both rosters were cleaned
    # and one is East while the other is West
    if not year_failed and len(year_results) == 2:

        conferences = {
            df["conference"].iloc[0]
            for df in year_results
        }

        if conferences == {"east", "west"}:
            cleaned_all_stars.extend(year_results)

        else:
            failed_all_stars.append({
                "year": year,
                "table": None,
                "error": f"Invalid conference pair: {conferences}"
            })


# Summary
print("Successfully cleaned tables:", len(cleaned_all_stars))
print("Expected tables:", 52)
print("Failures:", len(failed_all_stars))

# Display any problems
failure_report = pd.DataFrame(
    failed_all_stars,
    columns=["year", "table", "error"]
)

display(failure_report)
```

```python

# S14C — Concatenate and validate cleaned rosters

# 1. Combine all 52 cleaned DataFrames
all_stars = pd.concat(
    cleaned_all_stars,
    ignore_index=True
)

# 2. Basic dataset inspection
print("Shape:", all_stars.shape)
print("Years:", all_stars["year"].nunique())
print("Conferences:", all_stars["conference"].unique())

# 3. Validate roster counts by year and conference
roster_counts = (
    all_stars
    .groupby(["year", "conference", "selection_type"])
    .size()
    .unstack(fill_value=0)
)

display(roster_counts)

# 4. Check for missing values
print("\nMissing values:")
print(all_stars.isna().sum())

# 5. Check for duplicate player-year records
duplicates = all_stars[
    all_stars.duplicated(
        subset=["year", "player_name"],
        keep=False
    )
]

print("\nDuplicate player-year records:", len(duplicates))
display(duplicates)

# 6. Inspect the final DataFrame
display(all_stars.head(10))
```

```python

# S14C — Final validation

assert len(cleaned_all_stars) == 52
assert len(failed_all_stars) == 0

# 1. Combine all cleaned roster tables
all_stars = pd.concat(
    cleaned_all_stars,
    ignore_index=True
)

# 2. Standardize optional votes column
if "votes" not in all_stars.columns:
    all_stars["votes"] = pd.NA

# 3. Dataset summary
print("Rows:", len(all_stars))
print("Years:", all_stars["year"].nunique())
print(
    "Conference rosters:",
    all_stars.groupby(["year", "conference"]).ngroups
)

# 4. Check cleaned player positions only
valid_positions = {
    "G", "F", "C",
    "G/F", "F/G",
    "C/F", "F/C",
    "G/C", "C/G"
}

unexpected_positions = all_stars.loc[
    ~all_stars["position"].isin(valid_positions),
    ["year", "player_name", "position", "conference"]
]

print("\nUnexpected CLEANED positions:")
display(unexpected_positions)

# 5. Starter and reserve counts
print("\nStarters and reserves:")
display(
    all_stars.groupby(
        ["year", "conference", "selection_type"]
    ).size().unstack(fill_value=0)
)

# 6. Duplicate player-years
print("\nDuplicate player-years:")
display(
    all_stars[
        all_stars.duplicated(
            ["year", "player_name"],
            keep=False
        )
    ]
)

# 7. Missing values
print("\nMissing values:")
print(all_stars.isna().sum())

# 8. Verify Chris Bosh (2009) and Pau Gasol (2010)
print("\nPreviously flagged F/C players:")
display(
    all_stars.loc[
        (
            (all_stars["year"] == 2009)
            & (all_stars["player_name"] == "Chris Bosh")
        )
        |
        (
            (all_stars["year"] == 2010)
            & (all_stars["player_name"] == "Pau Gasol")
        ),
        ["year", "player_name", "position", "conference"]
    ]
)
```
