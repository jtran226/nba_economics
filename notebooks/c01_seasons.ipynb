```python
from pathlib import Path

PROJECT_ROOT = Path("/Users/jonathantran/Desktop/nba_economics")
STAGING_DIR = PROJECT_ROOT / "data" / "staging"

for path in sorted(STAGING_DIR.rglob("*")):
    if path.is_file():
        print(path.relative_to(STAGING_DIR))
```

```python
import pandas as pd

# S09: NBA salary cap, luxury tax, and apron history
s09 = pd.read_csv(
    STAGING_DIR / "salary_cap_apron/s09_salary_cap_history.csv"
)

print("SHAPE:", s09.shape)
print("COLUMNS:", s09.columns.tolist())

display(s09.head(5))
```

```python
import pandas as pd

s09 = pd.read_csv(
    STAGING_DIR / "salary_cap_apron/s09_salary_cap_history.csv"
)

s13b = pd.read_csv(
    STAGING_DIR / "CBA regime/s13_cba_regime_map.csv"
)

print("S09 columns:", s09.columns.tolist())
print("S13B columns:", s13b.columns.tolist())

display(s09.head(3))
display(s13b.head(3))
```

```python
# Confirm one row per season in each staged source.
assert s09["Season"].is_unique, "S09 contains duplicate seasons"
assert s13b["season"].is_unique, "S13B contains duplicate seasons"

# Confirm that the two datasets contain the same seasons.
assert set(s09["Season"]) == set(s13b["season"]), (
    "S09 and S13B have different season coverage"
)

print("Season keys are unique and coverage matches.")
```

```python
# Standardize S09's season column to match S13B.
s09_canonical = s09.rename(columns={"Season": "season"})

# Join the CBA identifier onto the salary-cap history.
c01_seasons = s09_canonical.merge(
    s13b[["season", "cba_id"]],
    on="season",
    how="left",
    validate="one_to_one"
)

# Inspect the result.
print("Shape:", c01_seasons.shape)
display(c01_seasons.head())
```

```python
# Inspect the column data types.
print(c01_seasons.dtypes)

# Validate the canonical table's structure.
assert len(c01_seasons) == 26, "Unexpected row count"
assert c01_seasons["season"].is_unique, "Duplicate seasons"
assert c01_seasons["season"].notna().all(), "Missing season IDs"
assert c01_seasons["cba_id"].notna().all(), "Missing CBA mappings"

# Check that the merge preserved all S09 records.
assert len(c01_seasons) == len(s09)

print("C01 structural validation passed.")
```

```python
print("Columns:", c01_seasons.columns.tolist())

print("\nMissing values by column:")
print(c01_seasons.isna().sum())

display(c01_seasons.tail(5))
```

```python
print(c01_seasons.dtypes)
```

```python
# Sort from earliest to latest season.
c01_seasons = (
    c01_seasons
    .sort_values("Year")
    .reset_index(drop=True)
)

display(c01_seasons.head())
```

```python
CANONICAL_DIR = PROJECT_ROOT / "data" / "canonical"
CANONICAL_DIR.mkdir(parents=True, exist_ok=True)

c01_seasons.to_csv(
    CANONICAL_DIR / "seasons.csv",
    index=False
)

print("C01 seasons saved successfully.")
```
