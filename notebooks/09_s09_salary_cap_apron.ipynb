Bringing in salary cap, luxury tax, apron information as well as exceptions information and history...

```python
import pandas as pd

salary_cap_history_table = pd.read_csv("../data/raw/salary_cap_apron/Salary Cap History.csv")

salary_cap_history_table.info()
```

```python
salary_cap_history_table
```

```python
salary_cap_table = salary_cap_history_table[salary_cap_history_table["Year"] != 2027].copy()

salary_cap_table.head()
```

```python
salary_cap_table["Salary Cap"] = (
    salary_cap_table["Salary Cap"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Salary Cap Floor"] = (
    salary_cap_table["Salary Cap Floor"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Luxury Tax"] = (
    salary_cap_table["Luxury Tax"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["1st Apron"] = (
    salary_cap_table["1st Apron"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["2nd Apron"] = (
    salary_cap_table["2nd Apron"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["BAE"] = (
    salary_cap_table["BAE"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Non-Taxpayer MLE"] = (
    salary_cap_table["Non-Taxpayer MLE"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Taxpayer MLE"] = (
    salary_cap_table["Taxpayer MLE"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Team Room MLE"] = (
    salary_cap_table["Team Room MLE"]
    .str.replace(r"[$,]", "", regex=True)
    .astype("Int64")
)

salary_cap_table["Salary Cap Floor Pct"] = (
    salary_cap_table["Salary Cap Floor Pct"]
    .str.replace("%", "", regex=False)
    .astype(float)
    / 100
)

salary_cap_table.head()
```

```python
salary_cap_table.info()
```

```python
import os

staging_dir = "../data/staging/salary_cap_apron"
os.makedirs(staging_dir, exist_ok=True)

salary_cap_table.to_csv(
    f"{staging_dir}/s09_salary_cap_history.csv",
    index=False
)
```
