```python
import pandas as pd
import numpy as np
import time
from nba_api.stats.endpoints import leaguedashplayerstats


def get_advanced_stats(year):

    advanced_stats_response = leaguedashplayerstats.LeagueDashPlayerStats(
    season=f"{year - 1}-{str(year)[-2:]}",
    measure_type_detailed_defense="Advanced",
    per_mode_detailed="PerGame"
)

    advanced_df = advanced_stats_response.get_data_frames()[0]

    advanced_df = advanced_df[[
    "PLAYER_ID",
    "PLAYER_NAME",
    "TEAM_ID",
    "TEAM_ABBREVIATION",
    "MIN",

    "OFF_RATING",
    "DEF_RATING",
    "NET_RATING",

    "AST_PCT",
    "AST_TO",
    "AST_RATIO",

    "OREB_PCT",
    "DREB_PCT",
    "REB_PCT",

    "TM_TOV_PCT",

    "EFG_PCT",
    "TS_PCT",
    "USG_PCT",

    "PACE",
    "PIE",
    "POSS"]]

    advanced_df["season"] = f"{year-1}-{year}"
    advanced_df["year"] = year

    return advanced_df

all_advanced_stats = []

for year in range(2001, 2027):

    advanced_df = get_advanced_stats(year)
    all_advanced_stats.append(advanced_df)

    print(f"{year}'s dataframe complete")
    time.sleep(3)

len(all_advanced_stats)
```

```python
all_advanced_stats[-1].isna().sum()
```

```python
advanced_stats = pd.concat(all_advanced_stats, ignore_index=True)

advanced_stats.tail()
```

```python
import os

os.makedirs(
    "../data/staging/advanced_player_stats",
    exist_ok=True
)

advanced_stats.to_csv("../data/staging/advanced_player_stats/s06_advanced_player_stats.csv",
    index=False
)
```
