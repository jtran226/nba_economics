##S13 - Setting up the pipeline to collect information on historical cba regime data and setting up proper mapping files...

```python
import pandas as pd
import os

cba_map = pd.read_csv("../data/raw/CBA_regime/CBA Mapping.csv")

cba_history = pd.read_csv("../data/raw/CBA_regime/CBA Regimes History.csv")
```

```python
print(cba_map.shape)
print(cba_history.shape)
```

```python
cba_map.head()
```

```python
# Check missing values
print(cba_map.isna().sum())
print(cba_history.isna().sum())

# Check duplicate records
print("Mapping duplicates:", cba_map.duplicated().sum())
print("History duplicates:", cba_history.duplicated().sum())

# Check CBA identifiers
print("Mapping IDs:", cba_map["cba_id"].unique())
print("History IDs:", cba_history["cba_id"].unique())
```

```python
cba_history["cba_id"] = (
    "cba_" +
    cba_history["cba_id"].str.extract(r"(\d{4})")[0]
)

cba_history.info()
```

```python
print(
    "Unmatched IDs:",
    set(cba_map["cba_id"]) - set(cba_history["cba_id"])
)

print("Duplicate seasons:", cba_map["season"].duplicated().sum())
print("Duplicate regimes:", cba_history["cba_id"].duplicated().sum())
```

```python
cba_history.head()
```

```python
print("Mapping shape:", cba_map.shape)
print("History shape:", cba_history.shape)

print("Duplicate seasons:", cba_map["season"].duplicated().sum())
print("Duplicate CBA IDs:", cba_history["cba_id"].duplicated().sum())

print(
    "Unmatched IDs:",
    set(cba_map["cba_id"]) - set(cba_history["cba_id"])
)
```

```python
cba_map.to_csv(
    "../data/staging/contracts/s13_cba_regime_map.csv",
    index=False
)

cba_history.to_csv(
    "../data/staging/contracts/s13_cba_regime_history.csv",
    index=False
)
```
