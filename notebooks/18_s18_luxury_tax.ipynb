```python
from pathlib import Path
import pandas as pd

raw_dir = Path("../data/raw/Luxury Tax")

# Select the 15 individual season files, excluding the historical table.
season_files = sorted(raw_dir.glob("* Lux Tax.csv"))

recent_tables = []

for file in season_files:
    df = pd.read_csv(file)

    # Remove completely empty rows.
    df = df.dropna(how="all")

    # Keep track of the original filename for validation.
    df["source_file"] = file.name

    recent_tables.append(df)

recent_combined = pd.concat(recent_tables, ignore_index=True)

print("Files loaded:", len(season_files))
print("Combined shape:", recent_combined.shape)

display(recent_combined.head())
```

```python
recent_clean = recent_combined.drop(
    columns=["Rank", "Record"]
).copy()

recent_clean = recent_clean.rename(columns={
    "Team": "team_id",
    "Year": "year",
    "Season": "season",
    "Tax Threshold": "tax_threshold",
    "Salaries": "taxable_payroll",
    "Space": "tax_space",
    "Tax BillEstimated": "estimated_tax_bill",
    "TotalTax Payroll + Est. Bill": "total_estimated_cost",
    "Active": "active_salary",
    "DeadEstimated": "dead_salary"
})

print(recent_clean.columns.tolist())
```

```python
financial_cols = [
    "tax_threshold",
    "taxable_payroll",
    "tax_space",
    "estimated_tax_bill",
    "total_estimated_cost",
    "active_salary",
    "dead_salary"
]

for col in financial_cols:
    recent_clean[col] = (
        recent_clean[col]
        .astype("string")
        .str.replace(r"[$,]", "", regex=True)
        .str.strip()
    )

    recent_clean[col] = pd.to_numeric(
        recent_clean[col],
        errors="coerce"
    )
```

```python
print("Shape:", recent_clean.shape)

print("\nData types:")
print(recent_clean.dtypes)

print("\nUnique seasons:", recent_clean["season"].nunique())
print("Unique teams:", recent_clean["team_id"].nunique())

print("\nDuplicate team-season rows:")
print(recent_clean.duplicated(subset=["team_id", "season"]).sum())

print("\nMissing financial values:")
print(recent_clean[financial_cols].isna().sum())

display(recent_clean.head())
```

```python
team_mapping = {
    "BKN": "BRK",
    "NJN": "BRK",
    "CHA": "CHO",
    "NOH": "NOP",
    "PHX": "PHO"
}

recent_clean["franchise_id"] = (
    recent_clean["team_id"].replace(team_mapping)
)
```

```python
checks = pd.DataFrame({
    "threshold_mismatch": (
        recent_clean["taxable_payroll"]
        + recent_clean["tax_space"]
        - recent_clean["tax_threshold"]
    ).abs() > 1,

    "total_cost_mismatch": (
        recent_clean["taxable_payroll"]
        + recent_clean["estimated_tax_bill"]
        - recent_clean["total_estimated_cost"]
    ).abs() > 1,

    "salary_breakdown_mismatch": (
        recent_clean["active_salary"]
        + recent_clean["dead_salary"].fillna(0)
        - recent_clean["taxable_payroll"]
    ).abs() > 1
})

print(checks.sum())
```

```python
historical = pd.read_csv(raw_dir / "Historical Tax Thresholds.csv")

# Promote the first row into column headers
historical.columns = historical.iloc[0]

# Remove that row from the data
historical = historical.iloc[1:].reset_index(drop=True)

print("Shape:", historical.shape)
print("Columns:", historical.columns.tolist())

display(historical.head(3))
```

```python
historical_long = historical.melt(
    id_vars="Team",
    value_vars=historical.columns[2:],
    var_name="year_threshold",
    value_name="tax_paid"
)

print(historical_long.shape)
display(historical_long.head())
```

```python
historical_long["tax_threshold"] = (
    historical_long["year_threshold"]
    .str[4:]                          # Remove the first 4 characters (YYYY)
    .str.replace("$", "", regex=False) # Remove dollar sign
    .str.replace("M", "", regex=False) # Remove M
    .astype(float)                    # Convert string to numeric
    * 1_000_000                       # Convert millions to dollars
)
```

```python
display(historical_long[["year_threshold", "tax_threshold"]].drop_duplicates().head())
```

```python
historical_long["year"] = (
    historical_long["year_threshold"]
    .str[:4]
    .astype(int)
)

historical_long["season"] = (
    historical_long["year"].astype(str)
    + "-"
    + (historical_long["year"] + 1).astype(str)
)
```

```python
historical_long.head()
```

```python
display(
    historical_long[
        ["year_threshold", "year", "season", "tax_threshold"]
    ].drop_duplicates().head(10)
)
```

```python
historical_long["tax_paid"] = (
    historical_long["tax_paid"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace("M", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

historical_long["tax_paid"] = pd.to_numeric(
    historical_long["tax_paid"],
    errors="coerce"
) * 1_000_000
```

```python
print(historical_long[["tax_threshold", "tax_paid"]].dtypes)

display(
    historical_long[
        ["Team", "season", "tax_threshold", "tax_paid"]
    ].head(10)
)
```

```python
historical_long = historical_long.rename(columns={
    "Team": "team_id"
})

historical_long["franchise_id"] = (
    historical_long["team_id"].replace(team_mapping)
)

print("Unique franchises:", historical_long["franchise_id"].nunique())

print(
    "Duplicate franchise-seasons:",
    historical_long.duplicated(["franchise_id", "season"]).sum()
)

print(
    historical_long.groupby("season")["franchise_id"].nunique()
)
```

```python
historical_long = historical_long.loc[
    ~(
        (historical_long["franchise_id"] == "CHO") &
        (historical_long["year"] < 2004)
    )
].copy()

print(historical_long.groupby("season")["franchise_id"].nunique().head(5))
print("Remaining rows:", len(historical_long))
```

```python
historical_long["tax_paid"] = historical_long["tax_paid"].fillna(0)

print("Missing tax payments:", historical_long["tax_paid"].isna().sum())
print("Teams with positive tax payments:",
      (historical_long["tax_paid"] > 0).sum())

display(
    historical_long[
        ["franchise_id", "season", "tax_threshold", "tax_paid"]
    ].head(10)
)
```

```python
# Standardize S18 to YYYY-YYYY
recent_clean["season"] = (
    recent_clean["season"].str[:5]
    + "20"
    + recent_clean["season"].str[-2:]
)

# Standardize the historical S18 table as well
historical_long["season"] = (
    historical_long["year"].astype(str)
    + "-"
    + (historical_long["year"] + 1).astype(str)
)

comparison = recent_clean[
    ["franchise_id", "season", "taxable_payroll"]
].merge(
    s08[["franchise_id", "Season", "Team_Payroll"]].rename(
        columns={"Season": "season"}
    ),
    on=["franchise_id", "season"],
    how="inner",
    validate="one_to_one"
)

comparison["payroll_difference"] = (
    comparison["Team_Payroll"] - comparison["taxable_payroll"]
)

print("Matched franchise-seasons:", len(comparison))
display(comparison.sample(n=5, random_state=42))
```

```python
comparison["difference_pct"] = (
    comparison["payroll_difference"]
    / comparison["taxable_payroll"] * 100
)

print(comparison["difference_pct"].describe().round(2))

print(
    "S08 higher:",
    (comparison["payroll_difference"] > 0).sum()
)

print(
    "Exact matches:",
    (comparison["payroll_difference"] == 0).sum()
)

print(
    "Within 2%:",
    (comparison["difference_pct"].abs() <= 2).sum()
)
```
