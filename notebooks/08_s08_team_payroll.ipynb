Building the s08 pipeline where we capture data that allows us to get team payroll historical information on the books by team per season...

```python
import pandas as pd

payroll = pd.read_csv("../data/raw/team_payroll/Historical NBA Team Payroll.csv")

payroll.shape
```

```python
payroll["Year"].isna().sum()
```

```python
cleaner_payroll = payroll[payroll["Season"].notna()].copy()

cleaner_payroll.info()

```

```python
cleaner_payroll = cleaner_payroll.rename(columns={
    "Team": "Team_Name",
    "Total": "Team_Payroll",
    "Cap Hit": "Salary_Cap_Hit",
    "Cap Space": "Salary_Cap_Space"
})

cleaner_payroll.head()
```

```python
cleaner_payroll["Team_Payroll"] = cleaner_payroll["Team_Payroll"].str.replace(r"[$,]","", regex=True).astype(int)
cleaner_payroll["Salary_Cap_Hit"] = cleaner_payroll["Salary_Cap_Hit"].str.replace(r"[$,]","",regex=True).astype(int)
```

```python
cleaner_payroll["Salary_Cap_Space"] = cleaner_payroll["Salary_Cap_Space"].str.replace(r"[$,]","",regex=True).replace("-", pd.NA).astype("Int64")

cleaner_payroll.head()
```

```python
cleaner_payroll.info()
```

```python
cleaner_payroll["Year"] = cleaner_payroll["Year"].astype(int)

cleaner_payroll.info()
```

```python
cleaner_payroll.isna().sum()
```

```python
import os

staging_dir = "../data/staging/team_payroll"
os.makedirs(staging_dir, exist_ok=True)

cleaner_payroll.to_csv(
    f"{staging_dir}/s08_team_payroll.csv",
    index=False
)
```
