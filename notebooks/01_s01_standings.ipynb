```python
def Conf_Standings(year):

    # 1. Build the URL for the requested season
    url = f"https://www.basketball-reference.com/leagues/NBA_{year}_standings.html"

    # 2. Download the HTML
    context = ssl.create_default_context(cafile=certifi.where())

    with urllib.request.urlopen(url, context=context) as response:
        html = response.read()

    # 3. Turn the HTML tables into a list of DataFrames
    tables = pd.read_html(io.BytesIO(html))

    # 4. Take the East and West standings tables
    east = tables[0].copy()
    west = tables[1].copy()

    # 5. Keep only actual team rows
    east = east[
        pd.to_numeric(east["W"], errors="coerce").notna()
    ].copy()

    west = west[
        pd.to_numeric(west["W"], errors="coerce").notna()
    ].copy()

    # 6. Standardize the team column
    east = east.rename(columns={"Eastern Conference": "team"})
    west = west.rename(columns={"Western Conference": "team"})

    # 7. Add conference
    east["conference"] = "East"
    west["conference"] = "West"

    # 8. Combine East and West
    standings = pd.concat(
        [east, west],
        ignore_index=True
    )

    # 9. Add season identifiers
    standings["season"] = f"{year - 1}-{year}"
    standings["year"] = year

    # 10. Return the finished DataFrame
    return standings


Conf_Standings(2013)
Conf_Standings(2020)
```

```python
years = range(2001,2027)

standings_history = []

for year in years:

    season_standings = Conf_Standings(year)
    standings_history.append(season_standings)

historical_standings_df = pd.concat(
    standings_history,
    ignore_index=True
)
```
