```python
### one year test of the function

import time
import os 

def build_salaries_df():

    checkpoint_directory = "../data/raw/player_salaries"
    os.makedirs(checkpoint_directory, exist_ok=True)

    all_seasons = []

    for year in range(2001, 2027):

        all_pages = []

        for page in range(1, 21):

            url = f"https://www.espn.com/nba/salaries/_/year/{year}/page/{page}/seasontype/1"
            salaries = pd.read_html(url)
            salary_df = salaries[0]
    
            all_pages.append(salary_df)

            time.sleep(5)

        one_full_season = pd.concat(all_pages, ignore_index=True)

        one_full_season["season"] = f"{year - 1}-{year}"
        one_full_season["year"] = year

        one_full_season.to_csv(
            f"{checkpoint_directory}/s07_salaries_{year}.csv", index=False)
        
        all_seasons.append(one_full_season)

        print(f"{year}'s player salary data is now complete")

    all_salaries = pd.concat(all_seasons, ignore_index=True)

    return all_salaries 

player_salaries = build_salaries_df()





```

```python
player_salary_columns = [
    "rank",
    "player",
    "team",
    "salary",
    "season",
    "year"
]

player_salaries.columns = player_salary_columns
```

```python
cleaned_player_salaries["position"] = (
    cleaned_player_salaries["player"]
    .str.extract(r",\s*([^,]+)$")
)

```

```python
cleaned_player_salaries[
    ["player", "position"]
].head(20)
```

```python
cleaned_player_salaries["position"].value_counts(dropna=False)
```

```python
cleaned_player_salaries["player"] = (
    cleaned_player_salaries["player"]
    .str.replace(r",.*$", "", regex=True)
    .str.strip()
)

cleaned_player_salaries.head(25)
```

```python
cleaned_player_salaries["salary"] = (
    cleaned_player_salaries["salary"]
    .str.replace(r"[$,]", "", regex=True)
    .astype(int)
)

cleaned_player_salaries["rank"] = cleaned_player_salaries["rank"].astype(int)

cleaned_player_salaries.info()
```
