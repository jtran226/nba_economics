```python
import pandas as pd
import numpy as np
import time 
from nba_api.stats.endpoints import leaguedashplayerstats


def get_stats(year):

    player_stats_response = leaguedashplayerstats.LeagueDashPlayerStats(
    season=f"{year - 1}-{str(year)[-2:]}",
    measure_type_detailed_defense="Base",
    per_mode_detailed="PerGame"
)

    stats_df = player_stats_response.get_data_frames()[0]

    stats_df = stats_df[[
    "PLAYER_ID",
    "PLAYER_NAME",
    "TEAM_ID",
    "TEAM_ABBREVIATION",
    "AGE",
    "GP",
    "W",
    "L",
    "W_PCT",
    "MIN",
    "FGM",
    "FGA",
    "FG_PCT",
    "FG3M",
    "FG3A",
    "FG3_PCT",
    "FTM",
    "FTA",
    "FT_PCT",
    "OREB",
    "DREB",
    "REB",
    "AST",
    "TOV",
    "STL",
    "BLK",
    "PF",
    "PTS",
    "PLUS_MINUS"]]

    stats_df["season"] = f"{year - 1}-{year}"
    stats_df["year"] = year

    return stats_df

all_player_stats = []

for year in range(2001, 2027):

    stats_df = get_stats(year)
    all_player_stats.append(stats_df)

    print(f"Finished with the final {year} of dataframes")
    time.sleep(5)


len(all_player_stats) 

```

```python
all_player_stats[-1].columns
```

```python
player_stats = pd.concat(all_player_stats, ignore_index=True)

player_stats.isna().sum()
```

```python
import os

os.makedirs("../data/staging/player_stats", exist_ok=True)

player_stats.to_csv(
    "../data/staging/player_stats/s05_player_stats.csv",
    index=False
)
```
