## S02 Production Pipeline

```python
# Creating the final pipeline and automating away the code... #

import pandas as pd
import ssl
import certifi
import urllib.request
import io

def get_stats(year):

    url = f"https://www.basketball-reference.com/leagues/NBA_{year}.html"

    context = ssl.create_default_context(cafile=certifi.where())

    with urllib.request.urlopen(url, context=context) as response:
        html = response.read()

    tables = pd.read_html(io.BytesIO(html))

    if year <= 2015:
        advanced_stats = tables[8].copy()
    else:
        advanced_stats = tables[10].copy()

    return advanced_stats


def cleaned_stats(df, year):
    df = df.copy()

    new_columns = [
        "Rank",
        "Team",
        "Age",
        "W",
        "L",
        "PW",
        "PL",
        "MOV",
        "SOS",
        "SRS",
        "ORtg",
        "DRtg",
        "NRtg",
        "Pace",
        "FTr",
        "3PAr",
        "TS%",
        "separator_1",
        "off_eFG%",
        "off_TOV%",
        "off_ORB%",
        "off_FT/FGA",
        "separator_2",
        "def_eFG%",
        "def_TOV%",
        "def_DRB%",
        "def_FT/FGA",
        "separator_3",
        "Arena",
        "Attend.",
        "Attend./G"
    ]

    df.columns = new_columns

    df = df.drop(columns=["separator_1", "separator_2", "separator_3"]).copy()

    df = df[df["Team"] != "League Average"].copy()

    df["Team"] = (df["Team"].str.replace("*", "", regex=False))
    df["season"] = f"{year - 1}-{year}"
    df["year"] = year

    df["Rank"] = df["Rank"].astype(int)
    df["W"] = df["W"].astype(int)
    df["L"] = df["L"].astype(int)
    df["PW"] = df["PW"].astype(int)
    df["PL"] = df["PL"].astype(int)

    return df


import time

historical_advanced_stats = []

for year in range(2001, 2027):

    find_table = get_stats(year)

    cleaned_table = cleaned_stats(find_table, year)

    historical_advanced_stats.append(cleaned_table)


time.sleep(10)


len(historical_advanced_stats)

all_advanced_stats = pd.concat(historical_advanced_stats, ignore_index=True)


### SANITY CHECK FOR AFTER THE FUNCTIONS

# Basic structure
print(all_advanced_stats.shape)
all_advanced_stats.info()

# Historical coverage
print(all_advanced_stats["season"].nunique())
print(all_advanced_stats["year"].min())
print(all_advanced_stats["year"].max())

# Teams per season
print(
    all_advanced_stats.groupby("season")["Team"].nunique()
)

# Duplicate team-seasons
print(
    all_advanced_stats.duplicated(
        subset=["Team", "season"]
    ).sum()
)

# Missing values
print(
    all_advanced_stats.isna().sum()
)

# Make sure cleaning worked
print(
    (all_advanced_stats["Team"] == "League Average").sum()
)

print(
    all_advanced_stats["Team"].str.contains(
        "*",
        regex=False
    ).sum()
)




```
