```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path("/Users/jonathantran/Desktop/nba_economics")
STAGING_DIR = PROJECT_ROOT / "data" / "staging"
CANONICAL_DIR = PROJECT_ROOT / "data" / "canonical"

for source in ["s04", "s05", "s06"]:
    matches = sorted(
        p for p in STAGING_DIR.rglob("*.csv")
        if source in p.name.lower()
    )

    print(f"\n{source.upper()}:")
    for path in matches:
        print(" ", path.relative_to(STAGING_DIR))
```

```python
sources = {}

for source in ["s04", "s05", "s06"]:
    matches = [
        p for p in STAGING_DIR.rglob("*.csv")
        if source in p.name.lower()
    ]

    assert len(matches) == 1, (
        f"{source}: expected 1 file, found {len(matches)}"
    )

    sources[source] = pd.read_csv(matches[0])

s04 = sources["s04"]
s05 = sources["s05"]
s06 = sources["s06"]

for name, df in sources.items():
    print(f"\n{name.upper()}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Seasons:", df["season"].nunique() if "season" in df.columns else "Check season column")
    display(df.head(3))
```

```python
keys = ["PLAYER_ID", "TEAM_ID", "season"]

for name, df in sources.items():

    print(f"\n{'=' * 55}")
    print(name.upper())
    print(f"{'=' * 55}")

    print("Total rows:", len(df))

    print(
        "Unique player-team-seasons:",
        df[keys].drop_duplicates().shape[0]
    )

    print(
        "Duplicate player-team-season keys:",
        df.duplicated(keys).sum()
    )

    print(
        "Unique player-seasons:",
        df[["PLAYER_ID", "season"]]
        .drop_duplicates()
        .shape[0]
    )

    print(
        "Missing key values:",
        df[keys].isna().sum().to_dict()
    )

    print(
        "Unique players:",
        df["PLAYER_ID"].nunique()
    )

    print(
        "Unique teams:",
        df["TEAM_ID"].nunique()
    )
```

```python
player_team_counts = (
    s04.groupby(["PLAYER_ID", "season"])["TEAM_ID"]
    .nunique()
    .reset_index(name="team_count")
)

multi_team_players = player_team_counts[
    player_team_counts["team_count"] > 1
]

print(
    "Player-seasons involving multiple teams:",
    len(multi_team_players)
)

print(
    "Maximum teams for one player in one season:",
    player_team_counts["team_count"].max()
)

display(multi_team_players.head(10))
```

```python
keys = ["PLAYER_ID", "TEAM_ID", "season"]

for name in ["s05", "s06"]:

    comparison = s04[keys].merge(
        sources[name][keys],
        on=keys,
        how="outer",
        indicator=True,
        validate="one_to_one"
    )

    print(f"\nS04 vs {name.upper()}")
    print(comparison["_merge"].value_counts())
```

```python
# Compare overlapping S04 and S06 advanced statistics.

advanced_cols = [
    "NET_RATING",
    "OREB_PCT",
    "DREB_PCT",
    "USG_PCT",
    "TS_PCT",
    "AST_PCT"
]

keys = ["PLAYER_ID", "TEAM_ID", "season"]

advanced_check = s04[
    keys + advanced_cols
].merge(
    s06[keys + advanced_cols],
    on=keys,
    how="inner",
    validate="one_to_one",
    suffixes=("_s04", "_s06")
)

for col in advanced_cols:
    left = advanced_check[f"{col}_s04"]
    right = advanced_check[f"{col}_s06"]

    mismatches = (
        left.ne(right)
        & ~(left.isna() & right.isna())
    ).sum()

    print(f"{col}: {mismatches} mismatches")
```

```python
totals_check = s04[
    keys + ["GP", "PTS", "REB", "AST"]
].merge(
    s05[
        keys + ["GP", "PTS", "REB", "AST"]
    ],
    on=keys,
    validate="one_to_one",
    suffixes=("_total", "_per_game")
)

for stat in ["PTS", "REB", "AST"]:
    estimated_total = (
        totals_check[f"{stat}_per_game"]
        * totals_check["GP_per_game"]
    )

    difference = (
        totals_check[f"{stat}_total"] - estimated_total
    ).abs()

    print(
        f"{stat}:",
        "within 0.5 per game played:",
        (difference <= 0.5 * totals_check["GP_per_game"]).mean()
    )

print(
    "Games played mismatches:",
    (
        totals_check["GP_total"]
        != totals_check["GP_per_game"]
    ).sum()
)
```

```python
# C05: Player-season performance with recorded franchise assignment
# Sources: S04 roster, S05 traditional stats, S06 advanced stats

keys = ["PLAYER_ID", "TEAM_ID", "season"]

# 1. S04: identity, physical characteristics, and season totals.
# Biographical fields already live in C03, so we don't duplicate them here.
roster = s04[
    keys + [
        "AGE",
        "PLAYER_HEIGHT_INCHES",
        "PLAYER_WEIGHT",
        "GP",
        "PTS",
        "REB",
        "AST"
    ]
].rename(columns={
    "PLAYER_ID": "player_id",
    "TEAM_ID": "franchise_id",
    "AGE": "age",
    "PLAYER_HEIGHT_INCHES": "height_inches",
    "PLAYER_WEIGHT": "weight_lbs",
    "GP": "games_played",
    "PTS": "points_total",
    "REB": "rebounds_total",
    "AST": "assists_total"
})

# 2. S05: traditional performance statistics.
# MIN and scoring/rebounding statistics are per-game averages.
traditional = s05[
    keys + [
        "W", "L", "W_PCT", "MIN",
        "FGM", "FGA", "FG_PCT",
        "FG3M", "FG3A", "FG3_PCT",
        "FTM", "FTA", "FT_PCT",
        "OREB", "DREB", "REB", "AST",
        "TOV", "STL", "BLK", "PF",
        "PTS", "PLUS_MINUS"
    ]
].rename(columns={
    "PLAYER_ID": "player_id",
    "TEAM_ID": "franchise_id",
    "W": "wins",
    "L": "losses",
    "W_PCT": "win_pct",
    "MIN": "minutes_per_game",
    "FGM": "field_goals_made_per_game",
    "FGA": "field_goals_attempted_per_game",
    "FG_PCT": "field_goal_pct",
    "FG3M": "three_pointers_made_per_game",
    "FG3A": "three_pointers_attempted_per_game",
    "FG3_PCT": "three_point_pct",
    "FTM": "free_throws_made_per_game",
    "FTA": "free_throws_attempted_per_game",
    "FT_PCT": "free_throw_pct",
    "OREB": "offensive_rebounds_per_game",
    "DREB": "defensive_rebounds_per_game",
    "REB": "rebounds_per_game",
    "AST": "assists_per_game",
    "TOV": "turnovers_per_game",
    "STL": "steals_per_game",
    "BLK": "blocks_per_game",
    "PF": "personal_fouls_per_game",
    "PTS": "points_per_game",
    "PLUS_MINUS": "plus_minus_per_game"
})

# 3. S06: advanced performance statistics.
# Overlapping S04 advanced fields were validated as identical.
advanced = s06[
    keys + [
        "OFF_RATING", "DEF_RATING", "NET_RATING",
        "AST_PCT", "AST_TO", "AST_RATIO",
        "OREB_PCT", "DREB_PCT", "REB_PCT",
        "TM_TOV_PCT", "EFG_PCT", "TS_PCT",
        "USG_PCT", "PACE", "PIE", "POSS"
    ]
].rename(columns={
    "PLAYER_ID": "player_id",
    "TEAM_ID": "franchise_id",
    "OFF_RATING": "offensive_rating",
    "DEF_RATING": "defensive_rating",
    "NET_RATING": "net_rating",
    "AST_PCT": "assist_pct",
    "AST_TO": "assist_to_turnover_ratio",
    "AST_RATIO": "assist_ratio",
    "OREB_PCT": "offensive_rebound_pct",
    "DREB_PCT": "defensive_rebound_pct",
    "REB_PCT": "rebound_pct",
    "TM_TOV_PCT": "team_turnover_pct",
    "EFG_PCT": "effective_field_goal_pct",
    "TS_PCT": "true_shooting_pct",
    "USG_PCT": "usage_pct",
    "PACE": "pace",
    "PIE": "player_impact_estimate",
    "POSS": "possessions"
})

# 4. Merge the three validated source tables.
canonical_keys = ["player_id", "franchise_id", "season"]

player_team_seasons = (
    roster
    .merge(
        traditional,
        on=canonical_keys,
        how="left",
        validate="one_to_one"
    )
    .merge(
        advanced,
        on=canonical_keys,
        how="left",
        validate="one_to_one"
    )
)

print("Shape:", player_team_seasons.shape)
display(player_team_seasons.head(3))
```

```python
# Load canonical reference tables.
players = pd.read_csv(CANONICAL_DIR / "players.csv")
aliases = pd.read_csv(
    CANONICAL_DIR / "franchise_season_aliases.csv"
)

# Primary-key checks.
key_cols = ["player_id", "franchise_id", "season"]

print("Total rows:", len(player_team_seasons))
print("Duplicate primary keys:", player_team_seasons.duplicated(key_cols).sum())
print("Duplicate player-seasons:",
      player_team_seasons.duplicated(["player_id", "season"]).sum())
print("Missing primary-key values:",
      player_team_seasons[key_cols].isna().sum().to_dict())

# Verify player identities against C03.
unmatched_players = ~player_team_seasons["player_id"].isin(
    players["player_id"]
)

print("Unmatched player IDs:", unmatched_players.sum())

# Verify franchise-season assignments against C02.
franchise_check = player_team_seasons[
    ["franchise_id", "season"]
].merge(
    aliases[["franchise_id", "season"]],
    on=["franchise_id", "season"],
    how="left",
    indicator=True,
    validate="many_to_one"
)

print("Franchise-season mapping:")
print(franchise_check["_merge"].value_counts())

# Check that the source joins produced no missing performance records.
print(
    "Missing traditional statistics:",
    player_team_seasons["games_played"].notna().sum()
    - player_team_seasons["points_per_game"].notna().sum()
)

print(
    "Missing advanced statistics:",
    player_team_seasons["offensive_rating"].isna().sum()
)

print("Seasons:", player_team_seasons["season"].nunique())
print("Columns:", len(player_team_seasons.columns))
```

```python
# Final assertions
assert len(player_team_seasons) == 12810
assert not player_team_seasons.duplicated(
    ["player_id", "franchise_id", "season"]
).any()

assert not player_team_seasons[
    ["player_id", "franchise_id", "season"]
].isna().any().any()

assert unmatched_players.sum() == 0
assert (franchise_check["_merge"] == "both").all()

# Export
CANONICAL_DIR.mkdir(parents=True, exist_ok=True)

output_path = CANONICAL_DIR / "player_team_seasons.csv"

player_team_seasons.to_csv(output_path, index=False)

print("C05 exported:", output_path)
print("Shape:", player_team_seasons.shape)
```
